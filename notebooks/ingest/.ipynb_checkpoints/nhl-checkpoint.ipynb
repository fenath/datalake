# Pré configuração de ambiente

```python
import requests
from dataclasses import fields, dataclass
from datetime import datetime, timedelta
from pyiceberg.catalog import load_catalog
from pyiceberg.schema import (
    Schema, 
    NestedField, 
    FloatType, 
    IntegerType,
    LongType,
    TimestampType, 
    StringType
)
import polars as pl
import duckdb
import json
import uuid


today = datetime.now()
first_of_month = datetime(today.year, today.month, 1)
today_str = today.strftime("%Y-%m-%d")
fom_str = first_of_month.strftime("%Y-%m-%d")

cat = load_catalog("dev")

TYPE_MAP = {
    str: StringType(),
    int: LongType(),
    float: FloatType(),
    datetime: TimestampType(),
}

def schema_from_dataclass(cls):
    iceberg_fields = []
    for i, f in enumerate(fields(cls), start=1):
        iceberg_fields.append(
            NestedField(
                i,
                f.name,
                TYPE_MAP[f.type],
                is_optional=True
            )
        )
    return Schema(*iceberg_fields)

def daterange(start_date, end_date, step=timedelta(days=1)):
    """
    A generator function that yields dates between start_date and end_date.
    The end_date is not included in the range.
    """
    current_date = start_date
    while current_date < end_date:
        yield current_date
        current_date += step

print(f'duckdb.version: {duckdb.__version__}')
```

```python
try:
    con.execute("""
        DETACH dev;
    """)
    
    if con is not None:
        con.close()
except:
    pass
con = duckdb.connect('duck.db')

con.sql("""
    INSTALL httpfs;
    LOAD httpfs;
    CREATE OR REPLACE SECRET minio (
        TYPE S3,
        KEY_ID 'minioadmin',
        SECRET 'minioadmin',
        ENDPOINT 'minio:9000', -- Remove http:// prefix here
        REGION 'us-east-1',
        USE_SSL false,
        URL_STYLE 'path'
    );
""")

con.sql("""
    INSTALL iceberg;
    LOAD iceberg;
    
    ATTACH 'dev' (
        TYPE iceberg
        , AUTHORIZATION_TYPE none
        , ENDPOINT 'http://rest:8181'
    );

    --show schemas 
""")
con.sql("show tables from dev.silver").pl()
```

```python
@dataclass
class NhlApiCall:
    requested_at: datetime
    endpoint: str
    status: int
    response: str

schema = schema_from_dataclass(NhlApiCall)
cat.create_table_if_not_exists("bronze.nhl_api_calls", schema=schema)

table = cat.load_table("bronze.nhl_api_calls")
df = pl.from_arrow(table.scan().to_arrow())
endpoint = "https://api-web.nhle.com/v1/schedule/{day}"
for dt in [datetime(today.year, today.month, i+1) for i in range(today.day)]:
    dt_str = dt.strftime("%Y-%m-%d")
    parsed_endpoint = endpoint.format(day=dt_str)
    if dt in (df.filter(
        (pl.col("status") == 200)
        & (pl.col("endpoint") == parsed_endpoint)
    )):
        continue
        
    response = requests.get(parsed_endpoint)
    df.extend(pl.DataFrame({
        "requested_at": datetime.now(),
        "endpoint": parsed_endpoint,
        "status": response.status_code,
        "response": response.text
    }))

df.write_iceberg("bronze.nhl_api_calls", catalog=catalog, mode="overwrite")
```

```python
con.sql("""
    CREATE TABLE IF NOT EXISTS dev.bronze.request_queue (
        request_id UUID PRIMARY KEY, -- Boa prática ter PK
        endpoint TEXT UNIQUE,        -- Aqui a mágica acontece
        status TEXT,
        created_at TIMESTAMP
    )
""")

def add_request_to_queue(endpoint):
    endpoints = [endpoint] if isinstance(endpoint, str) else endpoint
    df = pl.DataFrame([{
        'request_id': uuid.uuid4().bytes,
        'endpoint': endpoint,
        'status': 'pending',
        'created_at': datetime.now()
    } for endpoint in endpoints])
    tbl = cat.load_table('bronze.request_queue')
    schema = tbl.schema().as_arrow()
    arrow_df = df.to_arrow().cast(schema)
    tbl.append(arrow_df)
    # df.write_iceberg(
    #     cat.load_table('bronze.request_queue'),
    #     mode='append'
    # )

def process_request_queue():
    to_update = []
    df = con.sql("""
        SELECT 
            request_id, 
            endpoint
        FROM dev.bronze.request_queue WHERE status = 'pending'
        ORDER BY created_at 
        LIMIT 50
    """).pl()
    api_calls = cat.load_table('bronze.nhl_api_calls')
    schema = api_calls.schema().as_arrow()
    rows = []
    for request_id, endpoint in df.rows():
        response = requests.get(endpoint, timeout=30)
        to_update.append(request_id)
        rows.append({
            "requested_at": datetime.now(),
            "endpoint": endpoint,
            "status": response.status_code,
            "response": response.text    
        })
        
    if df.is_empty():
        return

    df_rows = pl.DataFrame(rows)
    df_rows.write_iceberg(api_calls, mode="append")
    con.execute("""UPDATE dev.bronze.request_queue 
            SET status='done' 
            WHERE request_id IN 
            (
            SELECT CAST(x AS UUID)
            FROM UNNEST(?) t(x)
            )""", [to_update])
    print(f"{len(to_update)} requests atualizados!")
    
```

# Parsear api -> Games Schedule

```python
# retorna as chamadas de api que ainda nao foram populadas para jogos
df = con.sql("""
select 
    distinct ON (api.endpoint) api.endpoint,
    api.requested_at,
    json_extract(response, '$.gameWeek')::JSON as game_week,
        (response->>'$.totalGames')::INT as total_games
from dev.bronze.nhl_api_calls api
where not exists (
    select 1
    from dev.silver.nhl_games g
    where g.endpoint = api.endpoint
)
AND api.endpoint LIKE 'https://api-web.nhle.com/v1/schedule/%'
AND api.status = 200
ORDER BY api.endpoint, api.requested_at DESC
""").pl()

df = (
    df
    .map_columns("game_week", lambda s: s.str.json_decode())
    .explode("game_week")
    .unnest("game_week")
    .explode("games")
    .unnest("games")
    .with_columns(
        pl.col("venue").struct.field("default").alias("venue"),
        pl.col("awayTeam").struct.field("commonName").struct.field("default").alias("away_common_name"),
        pl.col("awayTeam").struct.field("abbrev").alias("away_abbrev"),
        pl.col("awayTeam").struct.field("id").alias("away_id"),
        pl.col("homeTeam").struct.field("commonName").struct.field("default").alias("home_common_name"),
        pl.col("homeTeam").struct.field("abbrev").alias("home_abbrev"),
        pl.col("homeTeam").struct.field("id").alias("home_id"),
        pl.col("endpoint").str.extract(r"/schedule/(\d{4}-\d{2}-\d{2})").str.to_date().alias("endpoint_date"),
        pl.col("date").str.to_date(),
        pl.col("startTimeUTC").str.to_datetime(time_zone="UTC"),
    )
    .drop(["numberOfGames", 'datePromo', 'neutralSite',
           'easternUTCOffset', 'venueUTCOffset', 'venueTimezone',
           'tvBroadcasts', "periodDescriptor", "gameOutcome", "winningGoalie", "winningGoalScorer",
           "threeMinRecap", "threeMinRecapFr", "condensedGame", "condensedGameFr",
           "gameCenterLink", "specialEvent", "ticketsLink", "ticketsLinkFr",
           "total_games", "awayTeam", "homeTeam",
          ], strict=False)
    .filter(pl.col("gameType") <= 3)
    .rename({
        "id": "game_id",
        "gameType": "game_type",
        'startTimeUTC': 'start_time_utc',
        'gameState': 'game_state',
        'gameScheduleState': 'game_schedule_state',
    })
    .select("game_id", "date", "venue", "start_time_utc", "game_state",
            "game_schedule_state", "away_id", "away_abbrev", "away_common_name", "home_id",
            "home_abbrev", "home_common_name", "endpoint", "endpoint_date", "requested_at",
           )
)

table = cat.load_table("silver.nhl_games")
df.write_iceberg(table, mode="append")
print(f"{df.height} Jogos adicionados ✅")
df = None
```

# Parsear play-by-play

```python
df_sem_pbp = con.sql("""select distinct on (g.game_id) g.game_id, date, requested_at, endpoint_date , game_state,
        (pbp.game_id IS NOT NULL) AS has_play_by_play
        from dev.silver.nhl_games  g
        LEFT JOIN (SELECT DISTINCT game_id FROM dev.silver.nhl_play_by_play) pbp 
            ON g.game_id = pbp.game_id
        where endpoint_date < today() 
            and date < today()
            and year(date) = year(today())
            and game_state in ('FINAL', 'OFF')
            --and has_play_by_play = false
        order by game_id, endpoint_date, requested_at
        """).pl()

con.register("tmp_no_pbp", df_sem_pbp)

con.sql("""
    CREATE OR REPLACE FUNCTION get_pbp_endpoint(game_id) AS 
    ('https://api-web.nhle.com/v1/gamecenter/' || game_id || '/play-by-play')
""")

df_pbp_responses = con.sql("""
    SELECT
        DISTINCT m.game_id,
        get_pbp_endpoint(m.game_id) as pbp_enpdpoint,
        api.endpoint,
        api.status,
        json_extract(api.response, '$.plays')::JSON as plays,
        json_extract(api.response, '$.gameDate')::TIMESTAMP as game_date,
        json_extract(api.response, '$.homeTeam.id')::JSON as home_team_id
    FROM tmp_no_pbp m
        LEFT JOIN dev.bronze.nhl_api_calls api
            ON api.endpoint = get_pbp_endpoint(m.game_id)
""").pl()
```

```python
# Listar jogos que ainda não possuem response na API
if (nao_requisitadas := df_pbp_responses.filter(
    (~(pl.col("status") == 200)) | (pl.col("status").is_null())
)).height > 0:
    add_request_to_queue(nao_requisitadas["pbp_enpdpoint"].to_list())
    print(f"{nao_requisitadas.height} partidas sem pbp adicionadas à fila")
```

```python
# Preparacao de play by play 
current_batch_id = str(uuid.uuid4())
def transform_pbp(df_responses, batch_id):
    if df_responses.height == 0:
        return pl.DataFrame()
    df_pbp = (
        df_responses.filter(pl.col('status') == 200)
        .select("game_id", "plays", "home_team_id","game_date")
        .map_columns("plays", lambda s: s.str.json_decode())
        .explode("plays")
        .unnest("plays")
        .with_columns(
            pl.lit(current_batch_id).alias("batch_id"),
            pl.col("periodDescriptor").struct.field("number").alias("period"),
            pl.col("home_team_id").cast(pl.Int64),
            pl.col("game_date").dt.replace_time_zone("UTC"),
            pl.col("details").struct.field("xCoord").alias("x_coord"),
            pl.col("details").struct.field("yCoord").alias("y_coord"),
            pl.col("details").struct.field("typeCode").alias("detail_type"),
            pl.col("details").struct.field("shootingPlayerId").alias("player_id"),
            pl.col("details").struct.field("eventOwnerTeamId").alias("event_owner_team_id"),
            pl.col("details").struct.field("zoneCode").alias("zone_code"),
        )
        .rename({
            "eventId": "event_id",
            "timeInPeriod": "time_period",
            "situationCode": "situation_code",
            "homeTeamDefendingSide": "home_team_defending_side",
            "typeCode": "type_code",
            "typeDescKey": "type_desc",
            "sortOrder": "sort_order",
        })
        .with_columns(
            pl.when(
                # Condição de Inversão para o Home Team
                ((pl.col("event_owner_team_id") == pl.col("home_team_id")) & 
                 (pl.col("home_team_defending_side") == "right")) |
                # Condição de Inversão para o Away Team
                ((pl.col("event_owner_team_id") != pl.col("home_team_id")) & 
                 (pl.col("home_team_defending_side") == "left"))
            )
            .then(pl.col("x_coord") * -1)
            .otherwise(pl.col("x_coord"))
            .alias("x_normalized"),
            pl.col("time_period").str.split_exact(":", 1).alias("time_split"),
        )
        .with_columns(
            pl.col("time_split").struct.field("field_0").alias("time_minute").cast(pl.Int64),
            pl.col("time_split").struct.field("field_1").alias("time_second").cast(pl.Int64)
        )
        .select(
            "batch_id", "game_id", "game_date", "event_id", "period", "time_period",
            "time_minute", "time_second",
            "situation_code", 'home_team_defending_side', 'type_code',
            'type_desc', 'sort_order', 'x_coord', 'y_coord', "x_normalized",
            'zone_code', 'event_owner_team_id', 
        )
    )
    return df_pbp

df_pbp = transform_pbp(df_pbp_responses, current_batch_id)
```

## Ingestao dos eventos na tabela

```python
def save_params(batch_id, param_dict):
    df = pl.DataFrame([{
        'ingestion_id': batch_id,
        'param_name': key,
        'param_value': value
        
    } for key, value in param_dict.items()])
    df.write_iceberg(
        cat.load_table('bronze.ingestion_params'),
        mode='append'
    )


if df_pbp.height > 0:
    df_pbp.write_iceberg(
        cat.load_table("silver.nhl_play_by_play"), 
        mode="overwrite"
    )
    
    pl.DataFrame({
        "batch_id": current_batch_id,
        "event_time": datetime.now(),
        "source_table": "bronze.nhl_api_calls",
        "target_table": "silver.nhl_play_by_play",
        "final_status": "SUCCESS",
        "event_type": "FINISHED",
    }).write_iceberg(
        cat.load_table("dev.bronze.ingestion"),
        mode="append"
    )
    
    save_params(current_batch_id, {
        "total_games": df_pbp.select("game_id").n_unique(),
        "execution_mode": "vectorized_batch"
    })
```

## Exploracao duckdb

```python
%%duck
select distinct on (pbp.game_id, pbp.period, pbp.time_period, pbp.sort_order)
    pbp.game_id,
pbp.batch_id,
    g.home_id, 
    pbp.event_owner_team_id as team_id, 
    pbp.period,
    left(pbp.time_period, 2)::INT as tm_minute,
    right(pbp.time_period,2)::INT as tm_second,
    pbp.home_team_defending_side as home_side,
    pbp.type_desc,
    CASE 
        WHEN event_owner_team_id = home_id 
             AND home_team_defending_side = 'left'
        THEN x_coord * -1
        ELSE x_coord
    END AS x_norm,
    pbp.x_coord,
    pbp.y_coord,
    pbp.zone_code,
    pbp.sort_order
    -- , pbp.* 
from dev.silver.nhl_play_by_play pbp
    join dev.silver.nhl_games g on (g.game_id = pbp.game_id)
where 
    pbp.x_coord is not null AND pbp.y_coord is not null AND
 pbp.game_id = '2025020950'

ORDER by game_id, period, tm_minute, tm_second, sort_order
```

## Pipeline

```python
df = con.sql("""
select distinct on (pbp.game_id, pbp.period, pbp.time_period, pbp.sort_order)
    pbp.game_id,
pbp.batch_id,
    g.home_id, 
    pbp.event_owner_team_id as team_id, 
    pbp.period,
    left(pbp.time_period, 2)::INT as tm_minute,
    right(pbp.time_period,2)::INT as tm_second,
    pbp.home_team_defending_side as home_side,
    pbp.type_desc,
    CASE
        WHEN event_owner_team_id = home_id AND home_team_defending_side = 'left'
            THEN x_coord * -1
        WHEN event_owner_team_id != home_id AND home_team_defending_side = 'right'
            THEN x_coord * -1
        ELSE x_coord
    END AS x_norm,
    pbp.x_coord,
    pbp.y_coord,
    pbp.zone_code,
    pbp.sort_order
    -- , pbp.* 
from dev.silver.nhl_play_by_play pbp
    join dev.silver.nhl_games g on (g.game_id = pbp.game_id)
where 
    pbp.x_coord is not null AND pbp.y_coord is not null AND
 pbp.game_id = '2025020950' and team_id = home_id
    -- AND pbp.type_desc LIKE '%shot%'

ORDER by game_id, period, tm_minute, tm_second, sort_order
""").df()

import matplotlib.pyplot as plt
import seaborn as sns
from mplhockey import rink

FT_TO_M = 0.3048

fig, ax = plt.subplots(figsize=(10,4))

rink = rink.NHLRink()
rink.draw(ax=ax)

print(f'limiares: \n x: {df["x_norm"].min()} - {df["x_norm"].max()}\n',
     f'y: {df["y_coord"].min()} - {df["y_coord"].max()}')

hb = ax.hexbin(
    df["x_norm"].to_numpy() *FT_TO_M, 
    df["y_coord"].to_numpy() *FT_TO_M,
    gridsize=35,
    cmap="Reds",
    mincnt=1,
    extent=[-100, 100, -42.5, 42.5]
)

plt.colorbar(hb, ax=ax)

plt.show()
```

# Enriquecimento: sumario do jogo

```python
df = con.sql("""select * from dev.silver.nhl_play_by_play 
    where YEAR(game_date) = 2026    
""").pl()
df = (
    df
    .with_columns(
        pl.col("game_date").cast(pl.Date),
    )
    .group_by("game_id", "event_owner_team_id", "game_date")
    .agg([
        pl.col("type_desc").filter(pl.col("type_desc") == "faceoff").count().alias("faceoffs_ganhos"),
        pl.col("type_desc").filter(pl.col("type_desc").is_in(["shot-on-goal", "goal"])).count().alias("chutes_ao_gol"),
        pl.col("type_desc").filter(pl.col("type_desc") == "goal").count().alias("gols"),
        pl.col("type_desc").filter(pl.col("type_desc") == "blocked-shot").count().alias("bloqueado"),
        pl.col("type_desc").filter(pl.col("type_desc") == "takeaway").count().alias("desarmes"),
        pl.col("type_desc").filter(pl.col("type_desc") == "giveaway").count().alias("doacoes"),
        pl.col("type_desc").filter(pl.col("type_desc") == "hit").count().alias("rebatidas"),
    ])
    .filter(pl.col("event_owner_team_id").is_not_null())
    .sort(by=['game_id', 'event_owner_team_id'])
    .rename({
        "event_owner_team_id": "team_id"
    })
)
# df
```

### Criação da tabela

```python
from pyiceberg.partitioning import PartitionSpec, PartitionField
from pyiceberg.transforms import YearTransform


# 2. Cria a tabela passando o schema e o partition_spec
tbl = cat.create_table_if_not_exists(
    identifier="gold.nhl_game_team_stats", 
    schema=df.schema.to_arrow(),
)
try:
    # 2. Atualiza a partição para usar o ano do game_date
    # O Iceberg cuida de mapear o nome da coluna para o ID correto
    with tbl.update_spec() as update:
        update.add_field(
            source_column_name="game_date", 
            transform=YearTransform(), 
            partition_field_name="game_date_year"
        )
except:
    pass

# tbl
```

### Ingestão

```python
df.write_iceberg(tbl, mode='overwrite')
print(f"✅ atualizado tabela {tbl} com {df.height} registros")
```
